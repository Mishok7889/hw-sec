# Troubleshooting

Recovery and quick checks for the Husky + CW312T-SAM4S labs. Run the cell for whatever broke -
each one is independent.

## Board stuck after a failed flash

If `cw.program_target(...)` errored or the target went silent, the SAM4S is usually in its SAM-BA
bootloader. Reflash it, then re-run your lab's setup and flash cells.

```python
import chipwhisperer as cw
cw.program_sam_firmware(hardware_type='cwhusky')   # board is in the bootloader - reflash it
```

## Check the Husky is connected

Confirms the scope comes up and `import chipwhisperer` works (if the import fails, you picked the
wrong kernel).

```python
import chipwhisperer as cw
scope = cw.scope()
print(scope)
scope.dis()
```
