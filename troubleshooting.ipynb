# Troubleshooting

Recovery, firmware, and quick checks for the Husky + CW312T-SAM4S - not tied to any one week's lab.
Run the cell for whatever broke; each one is independent.

## Check the Husky is connected

Confirms the scope comes up and `import chipwhisperer` works (if the import fails, you picked the
wrong kernel).

```python
import chipwhisperer as cw
scope = cw.scope()
print(scope)
scope.dis()
```

## Update the Husky firmware

Update the capture hardware's own firmware, e.g. when a notebook warns it is out of date. It enters
the bootloader, flashes, and resets the board; reconnect with `cw.scope()` afterwards.

```python
import chipwhisperer as cw
scope = cw.scope()
scope.upgrade_firmware()   # update the Husky's own firmware, then reconnect with cw.scope()
```

## Board stuck after a failed flash

If `cw.program_target(...)` errored or the target went silent, the SAM4S is usually in its SAM-BA
bootloader. Reflash it, then re-run your lab's setup and flash cells.

```python
import chipwhisperer as cw
cw.program_sam_firmware(hardware_type='cwhusky')   # board is in the bootloader - reflash it
```
